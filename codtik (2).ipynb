```python
# "Si Jadwal Pintar" - Untuk Mengatur Jadwal Tugas Sekolah dengan Mudah
# Jalankan kode ini langsung, dan Anda bisa mengisi jadwal tugas sekolah tanpa ribet.
# Di akhir "done", Anda akan mendapat saran kapan harus mengerjakan tugas agar tidak bingung.
# Selamat mencoba!

import datetime

class SmartSchedule:
    def __init__(self):
        self.tasks = []  # List untuk menyimpan tugas: (tanggal_deadline, mata_pelajaran, deskripsi, prioritas, estimasi_jam)

    def add_task(self, deadline, subject, description, priority, est_hours):
        """Menambah tugas sekolah ke jadwal."""
        try:
            task_date = datetime.datetime.strptime(deadline, "%Y-%m-%d").date()
            self.tasks.append((task_date, subject, description, priority, est_hours))
            print(f"Tugas ditambahkan: {description} ({subject}) - Deadline: {deadline}, Prioritas: {priority}, Estimasi: {est_hours} jam")
        except ValueError:
            print("Format tanggal salah. Gunakan YYYY-MM-DD untuk deadline.")

    def view_schedule(self):
        """Melihat semua tugas yang diatur berdasarkan deadline dan prioritas."""
        if not self.tasks:
            print("Tidak ada tugas.")
        else:
            print("\nJadwal Tugas Sekolah Anda:")
            # Urutkan berdasarkan deadline (terdekat dulu), lalu prioritas (tinggi dulu)
            sorted_tasks = sorted(self.tasks, key=lambda x: (x[0], -x[3]))  # Prioritas tinggi = angka besar
            for task in sorted_tasks:
                print(f"- Deadline: {task[0]}, Mata Pelajaran: {task[1]}, Tugas: {task[2]}, Prioritas: {task[3]}, Estimasi: {task[4]} jam")

    def remove_task(self, description):
        """Menghapus tugas berdasarkan deskripsi."""
        for task in self.tasks:
            if task[2] == description:
                self.tasks.remove(task)
                print(f"Tugas '{description}' dihapus.")
                return
        print(f"Tugas '{description}' tidak ditemukan.")

    def suggest_schedule(self):
        """Memberikan saran kapan harus mengerjakan tugas berdasarkan deadline dan prioritas."""
        if not self.tasks:
            print("Tidak ada tugas untuk disarankan.")
            return

        today = datetime.date.today()
        suggestions = []

        # Urutkan tugas berdasarkan deadline dan prioritas
        sorted_tasks = sorted(self.tasks, key=lambda x: (x[0], -x[3]))

        print("\nSaran Pengaturan Waktu untuk Tugas Sekolah:")
        print("Berdasarkan deadline dan prioritas, berikut saran kapan mulai mengerjakan agar tidak bingung:")

        for task in sorted_tasks:
            deadline, subject, desc, priority, est_hours = task
            days_left = (deadline - today).days

            if days_left < 0:
                suggestions.append(f"- {desc} ({subject}): Deadline sudah lewat! Segera kerjakan jika belum selesai.")
            elif days_left == 0:
                suggestions.append(f"- {desc} ({subject}): Kerjakan hari ini! Estimasi {est_hours} jam.")
            elif days_left <= 3:
                suggestions.append(f"- {desc} ({subject}): Mulai kerjakan dalam {days_left} hari. Bagi waktu {est_hours} jam per hari.")
            else:
                daily_hours = max(1, est_hours // days_left)  # Minimal 1 jam per hari
                suggestions.append(f"- {desc} ({subject}): Kerjakan {daily_hours} jam per hari selama {days_left} hari.")

        for sug in suggestions:
            print(sug)

        print("\nTips: Mulai dari prioritas tinggi, alokasikan waktu belajar harian, dan istirahat cukup agar tidak bingung.")

# Inisialisasi jadwal tugas
schedule = SmartSchedule()

# Loop interaktif untuk mengisi tugas sekolah
print("Selamat datang di Si Jadwal Pintar untuk Tugas Sekolah!")
print("Jika Anda bingung dengan jadwal tugas, isi langsung di sini.")
print("Prioritas: 1 (Rendah), 2 (Sedang), 3 (Tinggi).")
print("Ketik 'done' untuk selesai dan lihat hasil akhir dengan saran waktu.")

while True:
    action = input("\nPilih aksi (add/view/remove/done): ").strip().lower()
    if action == "add":
        deadline = input("Masukkan deadline (YYYY-MM-DD, contoh: 2023-10-05): ").strip()
        subject = input("Masukkan mata pelajaran (contoh: Matematika): ").strip()
        desc = input("Masukkan deskripsi tugas (contoh: Kerjakan soal halaman 10): ").strip()
        priority = input("Masukkan prioritas (1-3): ").strip()
        est_hours = input("Masukkan estimasi waktu (dalam jam, contoh: 2): ").strip()
        if priority in ['1', '2', '3'] and est_hours.isdigit():
            schedule.add_task(deadline, subject, desc, int(priority), int(est_hours))
        else:
            print("Prioritas harus 1-3, dan estimasi waktu harus angka.")
    elif action == "view":
        schedule.view_schedule()
    elif action == "remove":
        desc = input("Masukkan deskripsi tugas yang ingin dihapus: ").strip()
        schedule.remove_task(desc)
    elif action == "done":
        print("\nHasil Akhir Jadwal Tugas Sekolah Anda:")
        schedule.view_schedule()
        schedule.suggest_schedule()
        print("Terima kasih telah menggunakan Si Jadwal Pintar! Semoga membantu mengatur waktu dan mengurangi kebingungan.")
        break
    else:
        print("Aksi tidak valid. Coba lagi.")
```
